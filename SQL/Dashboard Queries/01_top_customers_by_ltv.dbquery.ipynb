```sql
-- ============================================================
-- QUERY 01 : Top customers by Lifetime Value
-- DASHBOARD : Bar chart — horizontal
-- X-axis    : name
-- Y-axis    : lifetime_value
-- Colour by : customer_tier
-- ============================================================

SELECT
    f.customer_key,
    c.name,
    c.city,
    c.city_tier,
    t.tier_name                             AS customer_tier,
    t.retention_strategy,
    ROUND(f.total_spend, 0)                 AS lifetime_value,
    f.txn_count                             AS transactions,
    ROUND(f.avg_order_value, 0)             AS avg_order_value,
    f.recency_days,
    f.churn_flag,
    ROUND(f.engagement_score, 1)            AS engagement_score,
    ROUND(f.support_health_score, 1)        AS support_health

FROM customer360_gold.gold_fact_customer_metrics  f
JOIN customer360_gold.gold_dim_customer           c ON f.customer_key = c.customer_key
JOIN customer360_gold.gold_dim_tier               t ON f.tier_key     = t.tier_key

WHERE f.txn_count > 0          -- exclude customers who never transacted
ORDER BY f.total_spend DESC
LIMIT 20;
```
