```sql
-- ============================================================
-- QUERY 03 : Revenue breakdown by segment and tier
-- DASHBOARD : Grouped bar chart
--             X-axis    : tier_name
--             Y-axis    : total_revenue
--             Group by  : segment_name
-- Also use as: Pie chart for revenue_share_pct by tier
-- ============================================================

SELECT
    t.tier_name,
    t.priority_rank,
    s.segment_name,
    s.target_channel,
    COUNT(f.customer_key)                       AS customer_count,
    ROUND(SUM(f.total_spend), 0)                AS total_revenue,
    ROUND(AVG(f.total_spend), 0)                AS avg_ltv,
    ROUND(AVG(f.rfm_score), 1)                  AS avg_rfm_score,
    ROUND(AVG(f.engagement_score), 1)           AS avg_engagement,
    ROUND(AVG(f.support_health_score), 1)       AS avg_support_health,
    SUM(f.churn_flag)                           AS churned_customers,
    ROUND(SUM(f.churn_flag) * 100.0
          / COUNT(f.customer_key), 1)           AS churn_rate_pct,
    ROUND(SUM(f.total_spend) * 100.0
          / SUM(SUM(f.total_spend)) OVER (), 1) AS revenue_share_pct

FROM customer360_gold.gold_fact_customer_metrics  f
JOIN customer360_gold.gold_dim_tier               t ON f.tier_key     = t.tier_key
JOIN customer360_gold.gold_dim_segment            s ON f.segment_key  = s.segment_key

GROUP BY
    t.tier_name,
    t.priority_rank,
    s.segment_name,
    s.target_channel

ORDER BY t.priority_rank, total_revenue DESC;
```
