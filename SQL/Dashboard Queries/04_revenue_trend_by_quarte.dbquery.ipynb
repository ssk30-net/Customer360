```sql
-- ============================================================
-- QUERY 04 : Revenue trend over time (quarterly)
-- DASHBOARD : Line chart
--             X-axis : fiscal_quarter_label
--             Y-axis : net_revenue  (primary)
--                      txn_count    (secondary axis)
-- ============================================================

SELECT
    d.fiscal_quarter_label,
    d.year,
    d.quarter,
    COUNT(ft.transaction_key)               AS txn_count,
    ROUND(SUM(ft.net_amount), 0)            AS net_revenue,
    ROUND(AVG(ft.net_amount), 0)            AS avg_order_value,
    COUNT(DISTINCT ft.customer_key)         AS unique_customers,
    SUM(CASE WHEN ft.status = 'failed'
             THEN 1 ELSE 0 END)             AS failed_txns,
    ROUND(
        SUM(CASE WHEN ft.status = 'failed'
                 THEN 1 ELSE 0 END) * 100.0
        / COUNT(ft.transaction_key), 1
    )                                       AS failure_rate_pct

FROM customer360_silver.fact_transactions  ft
JOIN customer360_gold.gold_dim_date        d  ON ft.date_key = d.date_key

GROUP BY
    d.fiscal_quarter_label,
    d.year,
    d.quarter

ORDER BY d.year, d.quarter;
```
