```sql
-- ============================================================
-- QUERY 05 : Engagement breakdown by city tier and platform
-- DASHBOARD : Two visuals from one query
--   Visual A : Stacked bar — city_tier vs avg_engagement_score
--   Visual B : Donut      — platform share of total sessions
-- ============================================================

SELECT
    c.city_tier,
    c.city,
    c.state,
    c.region,
    COUNT(DISTINCT f.customer_key)          AS customers,
    ROUND(AVG(f.engagement_score), 1)       AS avg_engagement_score,
    ROUND(AVG(f.session_count), 1)          AS avg_sessions_per_customer,
    ROUND(AVG(f.avg_session_mins), 1)       AS avg_session_duration_mins,
    ROUND(AVG(f.bounce_rate), 1)            AS avg_bounce_rate_pct,
    ROUND(AVG(f.total_spend), 0)            AS avg_ltv,
    SUM(f.churn_flag)                       AS churned,
    ROUND(SUM(f.churn_flag) * 100.0
          / COUNT(f.customer_key), 1)       AS churn_rate_pct

FROM customer360_gold.gold_fact_customer_metrics  f
JOIN customer360_gold.gold_dim_customer           c ON f.customer_key = c.customer_key

GROUP BY
    c.city_tier,
    c.city,
    c.state,
    c.region

ORDER BY c.city_tier, avg_engagement_score DESC;
```
